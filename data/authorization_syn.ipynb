```python
RAW_INPUT_PATH = "synthetic_authorization_messy_15900.csv"

df = pd.read_csv(RAW_INPUT_PATH)

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
print("COLUMN NAMES:")
print(df.columns.tolist())

print("\nDATA TYPES:")
print(df.dtypes)
```

```python
print("MISSING VALUES (before cleaning):")
print(df.isnull().sum())

print("\nDUPLICATE ROWS (full-row, before cleaning):")
print(df.duplicated().sum())

print("\nDATASET INFO:")
df.info()
```

```python
# Reveal formatting inconsistency: same logical category, different text
categorical_cols = [
    "urgency", "authorization_status", "submission_channel",
    "authorization_type", "place_of_service", "service_type", "specialty"
]

for col in categorical_cols:
    print(f"\n{col} — raw unique values:")
    print(df[col].dropna().unique())
```

```python
# Reveal ID whitespace / stray formatting
id_cols = ["authorization_id", "member_id", "provider_id", "facility_id"]

for col in id_cols:
    stripped_diff = (
        df[col].dropna().astype(str).str.strip()
        != df[col].dropna().astype(str)
    ).sum()
    print(f"{col}: {stripped_diff} values have leading/trailing whitespace")
```

```python
# Reveal mixed date formats
sample_dates = pd.concat([
    df["service_date"].dropna().sample(15, random_state=1),
    df["authorization_date"].dropna().sample(15, random_state=1)
])
print("Sample raw date strings (note the mixed formats):")
print(sample_dates.tolist())
```

```python
# --- Step A: strip whitespace from every string/object column ---
string_cols = df.select_dtypes(include="object").columns

for col in string_cols:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

print("Whitespace stripped from all string columns.")
```

```python
# --- Step B: standardize casing on fixed-vocabulary categorical columns ---
# Maps any casing/spacing variant back to the canonical label (case-insensitive match).
# Values that don't match ANY known category are left as-is, so Part 2's
# rule_invalid_* checks can still correctly flag them as invalid categories.

canonical_categories = {
    "urgency": ["Routine", "Urgent", "Emergency"],
    "authorization_status": ["Approved", "Denied", "Pending"],
    "submission_channel": ["Portal", "EDI", "Fax", "Phone"],
    "authorization_type": ["Initial", "Extension", "Concurrent", "Retrospective"],
    "place_of_service": ["Clinic", "Hospital", "Home", "Emergency", "Outpatient"],
    "service_type": [
        "Inpatient", "MRI", "Surgery", "CT Scan", "DME",
        "Specialist Visit", "Physical Therapy", "Home Health"
    ],
    "specialty": [
        "Oncology", "General Medicine", "Orthopedics", "Radiology",
        "Neurology", "Surgery", "Cardiology", "Physical Therapy"
    ],
}

def normalize_category(value, valid_values):
    if pd.isna(value):
        return value
    lookup = {v.lower(): v for v in valid_values}
    return lookup.get(str(value).strip().lower(), value)

for col, valid_values in canonical_categories.items():
    before_unique = df[col].nunique(dropna=True)
    df[col] = df[col].apply(lambda x: normalize_category(x, valid_values))
    after_unique = df[col].nunique(dropna=True)
    print(f"{col}: {before_unique} unique values -> {after_unique} unique values")
```

```python
# --- Step C: parse mixed-format dates into proper datetime ---
# ISO, MM/DD/YYYY, and DD-MM-YYYY all resolve correctly.
# Genuinely unparseable strings (e.g. "unknown") become NaT (missing),
# which is the correct outcome - they ARE missing, not fixable.

def parse_mixed_date(value):
    if pd.isna(value):
        return pd.NaT
    for fmt in ("%Y-%m-%d", "%m/%d/%Y", "%d-%m-%Y"):
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.to_datetime(value, errors="coerce")  # last-resort flexible parse

for col in ["service_date", "authorization_date"]:
    before_missing = df[col].isnull().sum()
    df[col] = df[col].apply(parse_mixed_date)
    after_missing = df[col].isnull().sum()
    print(f"{col}: missing before={before_missing}, after parsing={after_missing} "
          f"(new NaT = genuinely unparseable strings, e.g. 'unknown')")
```

```python
# --- Step D: enforce numeric dtypes (coerce any stray non-numeric text) ---
numeric_cols = [
    "processing_time_hours", "missing_document_count", "resubmission_count",
    "appeal_flag", "claim_submitted_flag", "claim_amount"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Numeric dtypes enforced.")
print(df[numeric_cols].dtypes)
```

```python
# What was deliberately NOT touched (left for Part 2's rule engine):
# - malformed ID prefixes/case  -> rule_invalid_*_id
# - negative processing_time_hours / missing_document_count -> rule_negative_*
# - duplicate records -> rule_duplicate_record
# - future authorization dates -> rule_future_authorization_date

print("MISSING VALUES (after cleaning):")
print(df.isnull().sum())

print("\nConfirming untouched anomaly signals are still present:")
print("Negative processing_time_hours:", (df["processing_time_hours"] < 0).sum())
print("Negative missing_document_count:", (df["missing_document_count"] < 0).sum())
print("Full-row duplicates:", df.duplicated(subset=[c for c in df.columns if c != 'authorization_id']).sum())
print("Future authorization dates:", (df["authorization_date"] > pd.Timestamp.today()).sum())
```

```python
PREPROCESSED_OUTPUT_PATH = "authorization_preprocessed.csv"

df.to_csv(PREPROCESSED_OUTPUT_PATH, index=False)

print("Preprocessed data saved to:", PREPROCESSED_OUTPUT_PATH)
print("Shape:", df.shape)
```
